```python
#Algoritmo Bhaskara
import math
a=float(input("Digite o valor de a:"))
b=float(input("Digite o valor de b:"))
c=float(input("Digite o valor de c:"))
D=b**2 - 4*a*c
x1= (-(b) + math.sqrt(D))/2*a
x2= (-(b) - math.sqrt(D))/2*a
print(f"Raízes da equação: x1={x1} x2={x2}")
print(f"Valores de entrada: a:{a} b:{b} c:{c}")
```

```python
#Algoritmo Triangulo
A=int(input("Digite o valor do lado A:"))
B=int(input("Digite o valor do lado B:"))
C=int(input("Digite o valor do lado C:"))
if (A != 0) and (B != 0) and (C != 0):
    if (A < B + C) and (B < A + C) and (C < A + B):
        if (A == B) and (B == C):
            print("Triângulo Equilátero")
        elif (A == B) or (B == C) or (A == C):
            print("Triângulo Isósceles")
        else:
            print("Triângulo Escaleno")
    else:
        print("Os lados não formam um triângulo")
else:
    print("Os lados não formam um triângulo")
```

```python
# Algoritmo Produto
# From unittest import case  # ❌ Este import não é necessário aqui.
# O módulo 'unittest' é usado para testes automatizados em Python.
# Dentro dele existe uma função chamada 'case', mas ela não tem relação com o comando 'case' do 'match'.
# O 'match/case' é uma estrutura de controle nativa da linguagem, não precisa ser importada.
# Portanto, esta linha pode ser removida.

codigo = input("Digite o código do produto:")  # Solicita ao usuário o código do produto.

match codigo:  # Inicia uma estrutura de correspondência de padrões (pattern matching).
    case "1":  # Se o valor digitado for "1" (string), executa este bloco.
        print("Produto: Caderno")
    case "2":  # Se o valor for "2", executa este bloco.
        print("Produto: Lápis")
    case "3":  # Se o valor for "3", executa este bloco.
        print("Produto: Borracha")
    case _:  # O sublinhado (_) é o "default" — cobre qualquer outro valor.
        print("Diversos")

```

```python
# Algoritmo MediaPares
cont = 0        # contador de números pares
ACM = 0         # acumulador da soma dos números pares
numeros = []    # lista para armazenar todos os números digitados

# primeira entrada
N = int(input("Digite um número (ou 0 para sair): "))

while N != 0:
    numeros.append(N)   # guarda o número digitado na lista

    if N % 2 == 0:      # verifica se é par
        ACM += N        # soma ao acumulador
        cont += 1       # incrementa o contador

    N = int(input("Digite um número (ou 0 para sair): "))

# saída final
numeros.sort()  # ordena a lista em ordem crescente

# lista só dos pares
pares = [n for n in numeros if n % 2 == 0]

print(f"Números digitados (ordenados): {numeros}")
print(f"Números pares digitados: {pares}")

if cont > 0:
    media = ACM / cont
    print(f"A média dos números pares é: {media:.2f}")
    print(f"Total acumulado (ACM): {ACM}")
    print(f"Quantidade de números pares digitados (cont): {cont}")
else:
    print("Nenhum número par foi digitado.")
```

```python
# Algoritmo MediaPares
cont_pares = 0        # contador de números pares
ACM_pares = 0         # acumulador da soma dos números pares
cont_impares = 0      # contador de números ímpares
ACM_impares = 0       # acumulador da soma dos números ímpares
numeros = []          # lista para armazenar todos os números digitados

while True:
    entrada = input("Digite um número (ou 0 para sair): ")

    # trata entrada vazia
    if entrada.strip() == "":
        print("Entrada inválida! Digite um número inteiro.")
        continue

    try:
        N = int(entrada)
    except ValueError:
        print("Erro: digite apenas números inteiros.")
        continue

    if N == 0:
        break

    numeros.append(N)

    if N % 2 == 0:      # verifica se é par
        ACM_pares += N
        cont_pares += 1
    else:               # caso contrário, é ímpar
        ACM_impares += N
        cont_impares += 1

# saída final
numeros.sort()  # ordena a lista em ordem crescente

# listas separadas
pares = [n for n in numeros if n % 2 == 0]
impares = [n for n in numeros if n % 2 != 0]

print(f"\nNúmeros digitados (ordenados): {numeros}")
print(f"Números pares digitados: {pares}")
print(f"Números ímpares digitados: {impares}")

media_pares = None
media_impares = None

if cont_pares > 0:
    media_pares = ACM_pares / cont_pares
    print(f"A média dos números pares é: {media_pares:.2f}")
else:
    print("Nenhum número par foi digitado.")

if cont_impares > 0:
    media_impares = ACM_impares / cont_impares
    print(f"A média dos números ímpares é: {media_impares:.2f}")
else:
    print("Nenhum número ímpar foi digitado.")

# comparação entre médias
if media_pares is not None and media_impares is not None:
    if media_pares > media_impares:
        print("\nOs números pares tiveram a maior média.")
    elif media_impares > media_pares:
        print("\nOs números ímpares tiveram a maior média.")
    else:
        print("\nAs médias de pares e ímpares são iguais.")

# resumo final
print("\nResumo Final:")
if cont_pares > 0:
    print(f"- Pares: {cont_pares} números | Soma = {ACM_pares} | Média = {media_pares:.2f}")
else:
    print("- Nenhum número par digitado.")

if cont_impares > 0:
    print(f"- Ímpares: {cont_impares} números | Soma = {ACM_impares} | Média = {media_impares:.2f}")
else:
    print("- Nenhum número ímpar digitado.")
```

```python
# Algoritmo MediaParesImpares
import matplotlib.pyplot as plt

cont_pares = 0
ACM_pares = 0
cont_impares = 0
ACM_impares = 0
numeros = []

while True:
    entrada = input("Digite um número (ou 0 para sair): ")

    if entrada.strip() == "":
        print("Entrada inválida! Digite um número inteiro.")
        continue

    try:
        N = int(entrada)
    except ValueError:
        print("Erro: digite apenas números inteiros.")
        continue

    if N == 0:
        break

    numeros.append(N)

    if N % 2 == 0:
        ACM_pares += N
        cont_pares += 1
    else:
        ACM_impares += N
        cont_impares += 1

# saída final
numeros.sort()
pares = [n for n in numeros if n % 2 == 0]
impares = [n for n in numeros if n % 2 != 0]

print(f"\nNúmeros digitados (ordenados): {numeros}")
print(f"Números pares digitados: {pares}")
print(f"Números ímpares digitados: {impares}")

media_pares = None
media_impares = None

if cont_pares > 0:
    media_pares = ACM_pares / cont_pares
    print(f"A média dos números pares é: {media_pares:.2f}")
else:
    print("Nenhum número par foi digitado.")

if cont_impares > 0:
    media_impares = ACM_impares / cont_impares
    print(f"A média dos números ímpares é: {media_impares:.2f}")
else:
    print("Nenhum número ímpar foi digitado.")

# comparação
if media_pares is not None and media_impares is not None:
    if media_pares > media_impares:
        print("\nOs números pares tiveram a maior média.")
    elif media_impares > media_pares:
        print("\nOs números ímpares tiveram a maior média.")
    else:
        print("\nAs médias de pares e ímpares são iguais.")

# resumo final
print("\nResumo Final:")
if cont_pares > 0:
    print(f"- Pares: {cont_pares} números | Soma = {ACM_pares} | Média = {media_pares:.2f}")
else:
    print("- Nenhum número par digitado.")

if cont_impares > 0:
    print(f"- Ímpares: {cont_impares} números | Soma = {ACM_impares} | Média = {media_impares:.2f}")
else:
    print("- Nenhum número ímpar digitado.")

# gráfico comparativo com valores sobre as barras
if media_pares is not None or media_impares is not None:
    grupos = ["Pares", "Ímpares"]
    medias = [media_pares if media_pares else 0,
              media_impares if media_impares else 0]

    plt.bar(grupos, medias, color=["blue", "green"])
    plt.title("Comparação das Médias")
    plt.ylabel("Valor da Média")

    # adiciona os valores acima das barras
    for i, valor in enumerate(medias):
        plt.text(i, valor + 0.3, f"{valor:.2f}", ha='center', fontsize=10, fontweight='bold')

    plt.show()
```
